```python
import pandas as pd
import numpy as np

np.random.seed(42)
n = 200

df = pd.DataFrame({
    "customer_id": range(1001, 1001 + n),
    "age": np.random.randint(21, 70, n),
    "city": np.random.choice(["Mohali", "Delhi", "Pune", "Mumbai", "Chennai"], n),
    "account_type": np.random.choice(["Savings", "Current", "Salary"], n, p=[0.5, 0.2, 0.3]),
    "tenure_months": np.random.randint(1, 120, n),
    "balance": np.round(np.random.normal(50000, 30000, n), 2),
    "num_products": np.random.randint(1, 5, n),
    "monthly_txns": np.random.randint(0, 60, n),
    "credit_score": np.random.randint(450, 850, n),
    "has_credit_card": np.random.choice(["Yes", "No"], n),
    "churned": np.random.choice([0, 1], n, p=[0.8, 0.2]),
})

# dirty data on purpose (real work is messy)
df.loc[[5, 17, 42, 88], "balance"] = np.nan
df.loc[[10, 60], "credit_score"] = np.nan
df.loc[[3, 99], "city"] = " delhi "
df = pd.concat([df, df.iloc[[7, 20]]], ignore_index=True)   # 2 duplicate rows

df.to_csv("bank_churn.csv", index=False)
df.head(20)
```

```python
## Cleaning: fix city (strip spaces, title case). Fill missing balance with median of its account_type. Fill missing credit_score with overall median.

df['city'].unique()
```

```python
df['city']=df['city'].str.strip().str.title()
df['balance']=df['balance'].fillna(df.groupby('account_type')['balance'].transform("median"))
df['credit_score']=df['credit_score'].fillna(df['credit_score'].median())
```

```python
churned_rate = round(df['churned'].mean()*100,2)
print(churned_rate)

```

```python
def churn_by(col):
    return(df.groupby(col)['churned']
    .agg(customers="count", churned_rate_pct=lambda s: round(s.mean()*100 ,2))
    .sort_values('churned_rate_pct', ascending=False)
    .reset_index())

churn_by_city = churn_by("city")
churn_by_acct = churn_by("account_type")
print(churn_by_city)
print(churn_by_acct)

print('worst city:', churn_by_city.iloc[0]['city'], churn_by_city.iloc[0]['churned_rate_pct'] )
print('worst account type:', churn_by_acct.iloc[0]['account_type'], churn_by_acct.iloc[0]['churned_rate_pct'] )
```

```python
#  ---> Business segment: new column balance_band using pd.cut (Low < 20k, Medium 20k-60k, High > 60k). Show customer count, avg credit score, 
#churn % per band.
```

```python



conditions = [
              (df['balance'] < 20000 ),
              (df['balance'] >= 20000 ) & (df['balance'] < 60000 ),
              (df['balance'] >= 60000 )
]

choices= ['low','medium','high']

balance_band = np.select(conditions, choices, default ='unknow')
print(balance_band)

result = (
    df.groupby('balance_band', observed=False)
    .agg(
        customer_count=('customer_id', 'count'),
        avg_credit_score=('credit_score', 'mean'),
        churn_pct=('churned', lambda s: round(s.mean() * 100, 2))
    )
    .reset_index()
)

print(result)
```

```python
# ----> “Top 10 customers at risk of leaving.” Definition: not churned yet, monthly_txns < 10, balance above overall median. 
   ## Return customer_id, city, balance, monthly_txns, sorted by balance desc.
```

```python
median_balance = df["balance"].median()
print(median_balance)
```

```python
at_risk = df[
    (df["churned"] == 0) &
    (df["monthly_txns"] < 10) &
    (df["balance"] > median_balance)
]
print(at_risk)
```

```python
result = (
    at_risk[["customer_id", "city", "balance", "monthly_txns"]]
    .sort_values("balance", ascending=False)
    .head(10)
)

print(result)
```
